```python
import pandas as pd
pd.set_option('display.max_columns', None)
```

```python
df = pd.read_csv('dcorr_filtered_BBB_0.25.csv', low_memory=False)
```

```python
df.shape
```

```python
X_desc = df.drop(columns=['SMILES', 'BBB_Class'])
y = df['BBB_Class']
```

```python
X_desc.head(1)
```

```python
import numpy as np
def remove_highly_correlated_features(df, threshold=0.85, method='spearman'):

    corr_matrix = df.corr(method=method).abs()

    upper = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))

    to_drop = [column for column in upper.columns if any(upper[column] > threshold)]

    df_reduced = df.drop(columns=to_drop)

    return df_reduced, to_drop
```

```python
X_uncorrelated, dropped_features = remove_highly_correlated_features(X_desc, threshold=0.85, method='spearman')
print("Dropped features:", dropped_features)
```

```python
print(f"Original features (after variance filter): {X_desc.shape[1]}")
print(f"Remaining features (after correlation filter): {X_uncorrelated.shape[1]}")
print(f"Features removed due to high correlation: {len(dropped_features)}")
```

```python
X_uncorrelated.head(2)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X_uncorrelated, y, test_size=0.2, random_state=42, stratify=y)
```

```python
from sklearn.tree import DecisionTreeClassifier, plot_tree
import matplotlib.pyplot as plt


tree = DecisionTreeClassifier(max_depth=3, random_state=42)
tree.fit(X_train[['Lipinski', 'GhoseFilter']], y_train)


plt.figure(figsize=(12, 8))
plot_tree(
    tree,
    feature_names=['Lipinski', 'GhoseFilter'],
    class_names=['Not BBB permeable (-BBB)', 'BBB permeable (+BBB)'],
    filled=True,
    rounded=True,
    proportion=True,  
    precision=2,       
    fontsize=10
)
plt.title("Decision Tree on Rule-Based Features ", fontsize=14, weight='bold')
plt.show()
```

```python
df = X_train[['Lipinski','GhoseFilter']].copy()
df['y'] = y_train.values  

def rule_label(row):
    lip = 'Pass' if row['Lipinski'] > 0.5 else 'Fail'
    gho = 'Pass' if row['GhoseFilter'] > 0.5 else 'Fail'
    return f"{lip} Lipinski | {gho} Ghose"

df['rule'] = df.apply(rule_label, axis=1)

summary = (df.groupby('rule')
             .agg(Samples=('y','size'),
                  Positives=('y', lambda s: (s==1).sum() if s.dtype!='O' else (s=='+BBB').sum()))
             .reset_index())

summary['% +BBB'] = (summary['Positives'] / summary['Samples'] * 100).round(0).astype(int).astype(str) + '%'
summary = summary.rename(columns={'rule':'Rule Satisfaction', 'Positives':'# +BBB'})
summary = summary.sort_values('Samples', ascending=False)
print(summary[['Rule Satisfaction','Samples','% +BBB']])
```

```python
import seaborn as sns

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_uncorrelated)

plt.figure(figsize=(10,6))
sns.scatterplot(x=X_pca[:,0], y=X_pca[:,1], hue=y)
plt.title('PCA of Molecular Descriptors Colored by BBB Permeability')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.show()

loadings = pd.DataFrame(pca.components_.T, columns=['PC1', 'PC2'], index=X_uncorrelated.columns)
print("Top features influencing PC1:")
print(loadings['PC1'].abs().sort_values(ascending=False).head(10))
```
